### Importación de datos



```python
 import pandas as pd

 url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
 url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
 url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
 url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
# Calcular la facturación (suma de precios) para cada tienda
facturacion_1 = tienda['Precio'].sum()
facturacion_2 = tienda2['Precio'].sum()
facturacion_3 = tienda3['Precio'].sum()
facturacion_4 = tienda4['Precio'].sum()

# Imprimir los resultados
print(f"Facturación Tienda 1: ${facturacion_1}")
print(f"Facturación Tienda 2: ${facturacion_2}")
print(f"Facturación Tienda 3: ${facturacion_3}")
print(f"Facturación Tienda 4: ${facturacion_4}")

# Opcional: Si quieres saber el total de todas las tiendas juntas
total_general = facturacion_1 + facturacion_2 + facturacion_3 + facturacion_4
print(f"---")
print(f"Facturación Total Global: ${total_general}")
```

# 2. Ventas por categoría

```python
# Se Calcula y se cuentan las ventas por categoría para cada tienda
# value_counts() ordena automáticamente de la categoría más vendida a la menos vendida

print("--- Tienda 1: Ventas por Categoría ---")
print(tienda['Categoría del Producto'].value_counts())
print("\n")

print("--- Tienda 2: Ventas por Categoría ---")
print(tienda2['Categoría del Producto'].value_counts())
print("\n")

print("--- Tienda 3: Ventas por Categoría ---")
print(tienda3['Categoría del Producto'].value_counts())
print("\n")

print("--- Tienda 4: Ventas por Categoría ---")
print(tienda4['Categoría del Producto'].value_counts())
```

# 3. Calificación promedio de la tienda


```python
# Calcular la calificación promedio para cada tienda
promedio_1 = tienda['Calificación'].mean()
promedio_2 = tienda2['Calificación'].mean()
promedio_3 = tienda3['Calificación'].mean()
promedio_4 = tienda4['Calificación'].mean()

# Imprimir los resultados redondeados a 2 decimales para que se vea ordenado
print(f"Calificación Promedio Tienda 1: {promedio_1:.2f}")
print(f"Calificación Promedio Tienda 2: {promedio_2:.2f}")
print(f"Calificación Promedio Tienda 3: {promedio_3:.2f}")
print(f"Calificación Promedio Tienda 4: {promedio_4:.2f}")
```

# 4. Productos más y menos vendidos

```python
# Función para analizar y mostrar extremos de ventas
def mostrar_extremos(df, nombre_tienda):
    conteo = df['Producto'].value_counts()
    mas_vendido = conteo.index[0]
    menos_vendido = conteo.index[-1]

    print(f"--- {nombre_tienda} ---")
    print(f"Más vendido: {mas_vendido} ({conteo.iloc[0]} ventas)")
    print(f"Menos vendido: {menos_vendido} ({conteo.iloc[-1]} ventas)")
    print("")

# Ejecutar para las 4 tiendas
mostrar_extremos(tienda, "Tienda 1")
mostrar_extremos(tienda2, "Tienda 2")
mostrar_extremos(tienda3, "Tienda 3")
mostrar_extremos(tienda4, "Tienda 4")
```

# 5. Envío promedio por tienda



```python
# Calculo del costo de envío promedio para cada tienda
envio_promedio_1 = tienda['Costo de envío'].mean()
envio_promedio_2 = tienda2['Costo de envío'].mean()
envio_promedio_3 = tienda3['Costo de envío'].mean()
envio_promedio_4 = tienda4['Costo de envío'].mean()

# Se imprimen los resultados formateados
print(f"Envío Promedio Tienda 1: ${envio_promedio_1:.2f}")
print(f"Envío Promedio Tienda 2: ${envio_promedio_2:.2f}")
print(f"Envío Promedio Tienda 3: ${envio_promedio_3:.2f}")
print(f"Envío Promedio Tienda 4: ${envio_promedio_4:.2f}")
```

# Generando gráficos


```python
import matplotlib.pyplot as plt

# --- PREPARACIÓN DE DATOS ---
# Se crean listas con los resultados de tus ejercicios anteriores para graficar
tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
ingresos = [facturacion_1, facturacion_2, facturacion_3, facturacion_4]
envios_promedio = [envio_promedio_1, envio_promedio_2, envio_promedio_3, envio_promedio_4]
# Datos para el gráfico de pastel (Se usa Tienda 1 como ejemplo representativo)
categorias_t1 = tienda['Categoría del Producto'].value_counts()

# Configuración del tamaño de la figura (Ancho, Alto)
plt.figure(figsize=(18, 5))

# --- GRÁFICO 1: DE BARRAS (INGRESOS POR TIENDA) ---
plt.subplot(1, 3, 1) # (1 fila, 3 columnas, posición 1)
plt.bar(tiendas, ingresos, color=['#4CAF50', '#2196F3', '#FF9800', '#9C27B0'])
plt.title('Ingresos Totales por Tienda')
plt.ylabel('Monto ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# --- GRÁFICO 2: DE PASTEL (CATEGORÍAS - TIENDA 1) ---
plt.subplot(1, 3, 2) # (Posición 2)
plt.pie(categorias_t1, labels=categorias_t1.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribución de Ventas por Categoría (Tienda 1)')

# --- GRÁFICO 3: BARRAS HORIZONTALES (COSTO DE ENVÍO PROMEDIO) ---
plt.subplot(1, 3, 3) # (Posición 3)
plt.barh(tiendas, envios_promedio, color='#FF5722')
plt.title('Costo de Envío Promedio')
plt.xlabel('Costo ($)')
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Ajustar espacio entre gráficos y mostrar
plt.tight_layout()
plt.show()
```
